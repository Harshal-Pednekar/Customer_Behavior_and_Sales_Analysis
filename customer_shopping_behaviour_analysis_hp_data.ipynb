### 54 null values in 'review rating'

```python
csb['Review Rating'] = csb.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
csb.isnull().sum()
```

#### here we have imputed all the null values in 'Review rating' column

```python
# replacing current column names with snkae casing(lowercase+underscore) column names
csb.columns = csb.columns.str.lower() #to make it lowercase
csb.columns = csb.columns.str.replace(' ','_') #replacing space with underscore
```

```python
csb.columns
```

```python
#create a column age_group
labels = ['Young Adult', 'Adult', 'Middle Age', 'Senior']
csb['age_group'] = pd.qcut(csb['age'],q=4, labels = labels)
```

```python
csb[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

csb['purchase_frequency_days'] = csb['frequency_of_purchases'].map(frequency_mapping) #using map() to replace the text with the matching number in the column
```

```python
csb[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
csb[['discount_applied','promo_code_used']].head(10)
```

these two column seems to have same data

```python
(csb['discount_applied'] == csb['promo_code_used']).all()
```

#### The validation confirmed that whenever a discount was applied, the promo code status matched exactly. 
#### Since both columns store the same information, retaining only one avoids redundancy and simplifies the dataset.

```python
csb = csb.drop('promo_code_used', axis = 1)

# axis=0 (default) → Work down the rows (column-wise) //Go up and down (↓)
# axis=1 → Work across the columns (row-wise) //Go left to right (→)
```

```python
#step 1: connect to postgreSQL
username = "postgres" #default user
password = "root" #the pass you set during installation
host = "localhost"
post = "5432"
database = "customer_behavior_analysis" #the database you created in pgAdmin

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{post}/{database}")

#step 2: Load Dataframe into postgreSQL
table_name = "customer_data" #choose any table name
csb.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"data successfully loaded into table'{table_name}' in database '{database}'.")
```
